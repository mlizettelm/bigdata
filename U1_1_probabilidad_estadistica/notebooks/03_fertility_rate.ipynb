* Como podria calcular ahora las coincidencias usando Pandas?

### Otra forma de extraer los datos, es prepararlos desde el data frame de pandas

Ahora usaremos una funcion llamada merge()

```python
mexico.head(3)
```

```python
japon.head(3)
```

Creamos un nuevo data frame con las columnas que queremos 

```python
comparacion_mexico_japon = pd.merge(
    mexico[["year", "fertility_rate_hist"]],
    japon[["year", "fertility_rate_hist"]],
    on="year",
    suffixes=("_mexico", "_japon")
)
```

```python
comparacion_mexico_japon.head(3)
```

```python
comparacion_mexico_japon["diferencia"] = (
    comparacion_mexico_japon["fertility_rate_hist_mexico"]
    - comparacion_mexico_japon["fertility_rate_hist_japon"]
)
```

```python
comparacion_mexico_japon.head(3)
```

```python

plt.figure(figsize=(10, 5))

comparacion_mexico_japon.plot(
    x="year",
    y=["fertility_rate_hist_mexico", "fertility_rate_hist_japon","diferencia"],
    kind="line"
)

plt.xlabel("Año")
plt.ylabel("Tasa de fertilidad")
plt.title("Tasa de fertilidad: México vs Japón")
plt.grid(True)
plt.legend()
#plt.show()
```
